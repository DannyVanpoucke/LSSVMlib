```python
import numpy as np
import pandas as pd

from scipy.spatial.distance import cdist

from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

# Binary LSSVMCLassifier

```python
class _BinaryLSSVMClassifier(BaseEstimator, ClassifierMixin):
     def __init__(self,
                 gamma=1.0,
                 kernel="rbf",
                 c=1.0,
                 d=2,
                 sigma=1.0):

        self.gamma = gamma
        self.kernel = kernel

        self.c = c
        self.d = d
        self.sigma = sigma

        params = {}

        if kernel == "poly":
            params["c"] = c
            params["d"] = d

        elif kernel == "rbf":
            params["sigma"] = sigma

        self.kernel_ = self.__set_kernel(kernel, **params)

        self.x = None
        self.y = None

        self.coef_ = None
        self.intercept_ = None
# Binary fit 
def fit(self, X, y):

        self.x = np.asarray(X)
        self.y = np.asarray(y)

        self.__OptimizeParams()

        return self

# Binary predict
def predict(self, X):

        Ker = self.kernel_(X, self.x)

        score = np.dot(self.coef_, Ker.T) + self.intercept_

        return np.sign(score)
```
