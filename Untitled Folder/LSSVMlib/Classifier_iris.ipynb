```python
class LSSVMClassifier(BaseEstimator, ClassifierMixin):

    def __init__(self,
                 gamma: float = 1.0,
                 kernel: str = "rbf",
                 c: float = 1.0,
                 d: int = 2,
                 sigma: float = 1.0):

        self.gamma = gamma
        self.kernel = kernel

        self.c = c
        self.d = d
        self.sigma = sigma

        params = {}

        if self.kernel == "poly":
            params["c"] = self.c
            params["d"] = self.d

        elif self.kernel == "rbf":
            params["sigma"] = self.sigma

        self.kernel_ = self.__set_kernel(self.kernel, **params)

        # model parameters
        self.x = None
        self.y = None

        self.coef_ = None
        self.intercept_ = None

# Get params
    def get_params(self, deep=True):

        return {
            "gamma": self.gamma,
            "kernel": self.kernel,
            "c": self.c,
            "d": self.d,
            "sigma": self.sigma
        }

# Set params
    def set_params(self, **parameters):

        for parameter, value in parameters.items():
            setattr(self, parameter, value)

        params = {}

        if self.kernel == "poly":
            params["c"] = self.c
            params["d"] = self.d

        elif self.kernel == "rbf":
            params["sigma"] = self.sigma

        self.kernel_ = self.__set_kernel(self.kernel, **params)

        return self

# Kernels
    @staticmethod
    def __set_kernel(name, **params):

        def linear(xi, xj):
            return np.dot(xi, xj.T)


        def poly(xi, xj,
                 c=params.get("c", 1.0),
                 d=params.get("d", 2)):

            return ((np.dot(xi, xj.T) / c) + 1) ** d


        def rbf(xi, xj,
                sigma=params.get("sigma", 1.0)):

            if xi.ndim == 2 and xj.ndim == 2:

                return np.exp(
                    -cdist(xi, xj, metric="sqeuclidean")
                    / (2 * sigma**2)
                )

            ax = len(xj.shape) - 1

            return np.exp(
                -(np.dot(xi, xi)
                  + (xj**2).sum(axis=ax)
                  - 2 * np.dot(xi, xj.T))
                / (2 * sigma**2)
            )


        kernels = {
            "linear": linear,
            "poly": poly,
            "rbf": rbf
        }

        if name not in kernels:
            raise KeyError(f"Unknown kernel '{name}'")

        return kernels[name]

# Set_attributes letterlijk overnemen uit regressionklasse
    def set_attributes(self, **parameters):
        """
        Manually set the attributes of the model.

        Parameters:
        intercept_ : intercept
        coef_      : coefficients
        support_   : support vectors
        """

        for param, value in parameters.items():

            if param == "intercept_":
                self.intercept_ = value

            elif param == "coef_":
                self.coef_ = value

            elif param == "support_":
                self.x = value

# Optimize_params implementeren 
    def __OptimizeParams(self):
        """
        Solve the LS-SVM system for classification.
        """

    # Omega = y_i y_j K(x_i,x_j)
        Omega = np.multiply.outer(self.y, self.y) * self.kernel_(self.x, self.x)

        Ones = self.y.reshape(-1, 1)

        A = np.block([
        [np.array([[0]]), Ones.T],
        [Ones, Omega + (1/self.gamma) * np.eye(len(self.y))]
        ])

        B = np.concatenate(([0], np.ones(len(self.y))))

        solution = np.linalg.solve(A, B)

        self.intercept_ = solution[0]
        self.coef_ = solution[1:]

# fit() vrijwel identiek aan regression 
    def fit(self, X: np.ndarray, y: np.ndarray):
   
        if isinstance(X, (pd.DataFrame, pd.Series)): #checks if X is an instance of either types
            Xloc = X.to_numpy()
        else:
            Xloc = X

        if isinstance(y, (pd.DataFrame, pd.Series)):
            yloc = y.to_numpy()
        else:
            yloc = y
        labels = np.unique(yloc)

        if len(labels) != 2:
            raise ValueError(
                "Current implementation only supports binary classification."
            ) #extra controle 

        #check the dimensionality of the input
        if (Xloc.ndim == 2) and (yloc.ndim == 1):
            self.x = Xloc
            self.y = yloc
            self.__OptimizeParams()
        else:
            raise ValueError(
                "The fit procedure requires a 2D numpy array of features "\
                "and 1D array of targets"
            )
            
       return self 

        
#predict() 
    def predict(self, X):

        Ker = self.kernel_(X, self.x)

        score = np.dot(self.coef_, Ker.T) + self.intercept_

        return np.where(score >= 0, 1, -1)
```
