# 2. Data 

```python
pieken = pd.read_csv("CuNPs_Pieken.csv")
```

```python
pieken
```

## 2.1 Test/train split

```python
X = pieken[["Cu_conc[M]","Temperature[°C]","Time[min]"]]
y = pieken["Klasse"]

X_train, X_test, y_train, y_test = train_test_split(X,y, stratify=y, random_state=17, test_size=0.2, shuffle=True)
print('y_train : ')
print(y_train.head())
print('')
print('y_test : ')
print(y_test.head())
```

# 3. Hyper parameter tuning

```python
model = SVC(kernel='rbf',class_weight='balanced')
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## 3.1 Parameter range en Cross-Fold validation

```python
parameter_grid = {'C': [0.01, 0.1, 1, 10, 100],'gamma': [0.01, 0.1, 1, 10, 100]}

grid = GridSearchCV(model, parameter_grid, cv=3,scoring='recall')
grid.fit(X_train_scaled,y_train)
```

```python
# cross validation scores
scores = cross_val_score(model, X_train_scaled, y_train, cv=3)

print(scores)
print("Gemiddelde score:", scores.mean())
```

```python
print("best parameters", grid.best_params_)
```

```python
#best_model=grid.best_estimator_
#prediction=best_model.predict(X)
```

# 4. Training

```python
grid.best_params_['gamma']
```

```python
grid
```

```python
svm_clf = SVC(kernel='rbf', C=grid.best_params_['C'], gamma=grid.best_params_['gamma'], class_weight='balanced')#balanced toegevoegd
svm_clf.fit(X_train_scaled, y_train)
```

```python
# Predict met aangepaste threshold
y_scores = svm_clf.decision_function(X_test_scaled)
y_pred = (y_scores > -0.5).astype(int)

# Confusion matrix
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.show()
```

```python
test_accuracy=svm_clf.score(X_test_scaled,y_test) #overeenkomen van X-test en y_test
print(test_accuracy)
```
