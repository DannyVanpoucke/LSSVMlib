# 1. Imports

```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd 

#Sklearn bibliotheken
from sklearn import datasets
from matplotlib.ticker import (MultipleLocator, AutoMinorLocator)
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import r2_score, mean_squared_error #RMSE, MAE, r2 gebruiken
from sklearn.pipeline import Pipeline #niet heel belangrijk, zet samen standard scaler en linear svc
from sklearn.preprocessing import StandardScaler #standardisatie: gem 0, std 1 
from sklearn.svm import LinearSVC
from sklearn.svm import LinearSVR
from sklearn.model_selection import train_test_split
```

# 2. Data 

```python
pieken = pd.read_csv("CuNPs_Pieken.csv")
```

```python
pieken
```

## 2.1 Test/train split

```python
X = pieken[["Cu_conc[M]","Temperature[°C]","Time[min]"]]
y = pieken["Klasse"]

X_train, X_test, y_train, y_test = train_test_split(X,y , random_state=42, test_size=0.2, shuffle=True)
print('X_train : ')
print(X_train.head())
print('')
print('X_test : ')
print(X_test.head())
print('')
print('y_train : ')
print(y_train.head())
print('')
print('y_test : ')
print(y_test.head())
```

# 3. Hyper parameter tuning

```python
model = SVC(kernel='rbf')
```

## 3.1 Parameter range en Cross-Fold validation

```python
parameter_grid = {'C': [0.01, 0.1, 1, 10, 100],'gamma': [0.01, 0.1, 1, 10, 100]}

grid = GridSearchCV(model, parameter_grid, cv=3)
grid.fit(X_train,y_train)
```

```python
print("best parameters", grid.best_params_)
```

```python
best_model=grid.best_estimator_
prediction=best_model.predict(X)
```

# 4. Training

```python
svm_clf = Pipeline([("scaler", StandardScaler()),
("svc", SVC(kernel='rbf',gamma=0.01, C=0.01)),])
svm_clf.fit(X_train, y_train)
```

```python
test_accuracy=svm_clf.score(X_test,y_test) #overeenkomen van X-test en y_test
print(test_accuracy)
```

# 5. Test

```python
predicted=svm_clf.predict(X_test)
print(y_test) 
```

```python
test_accuracy=svm_clf.score(X_test,y_test) #overeenkomen van X-test en y_test
print(test_accuracy)
```
