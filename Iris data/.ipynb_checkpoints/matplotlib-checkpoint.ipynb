```python
import matplotlib.pyplot as plt

x=[0,1,2,3,4]
y=[0,1,4,9,16]

plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[0,1,2,3,4]
y=[0,1,4,9,16]

fig,ax=plt.subplots()
ax.plot(x, y, marker='o',label="Data Points")

ax.set_title("Basic Components of Matplotlib Figure")
ax.set_xlabel("X-Axis")
ax.set_ylabel("Y-Axis")

plt.show()
```

```python
# basic histograms 
import matplotlib.pyplot as plt
import numpy as np 
data=np.random.randn(1000) #generates 1000 random numbers from a standard normal distribution 
plt.hist(data,bins=30) #creates a histogram of the data with 30 bins (intervals
plt.show()
```

```python
# histogram with titles and labels 
import matplotlib.pyplot as plt
import numpy as np 

data=np.random.randn(1000) #generates 1000 random numbers from a standard normal distribution 

plt.hist(data,bins=30,edgecolor='black) #creates a histogram of the data with 30 bins (intervals)
plt.xlabel('title')
plt.ylabel('with')
plt.title('This is a histogram with title')
plt.show()
```
