```python
# oplossing 1 - 4 aparte histogrammen onder elkaar 
import matplotlib.pyplot as plt
import numpy as np 
from sklearn import datasets 

iris = datasets.load_iris()
X = iris["data"]
feature_names = iris["feature_names"]

plt.figure(figsize=(8, 8))

for i in range(4):
    plt.subplot(2, 2, i+1)
    plt.hist(X[:, i], bins=20, color='skyblue', edgecolor='navy')
    plt.title(feature_names[i])

plt.tight_layout()
plt.show()
```

```python
#oplossing 2 - elk histogram in aparte figuur 
for i in range(4):
    plt.figure()
    plt.hist(X[:, i], bins=20)
    plt.title(feature_names[i])
    plt.show()
```
